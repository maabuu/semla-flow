```python
df_db_meta = pd.read_csv("data/unconditional/drugbank/all_structure_links.csv")
db_approved_ids = set(df_db_meta[df_db_meta["Drug Groups"].str.contains("approved")]["DrugBank ID"])
```

```python
file = Path("data/unconditional/drugbank/all_structures_3d.sdf")
blocks = open(file).read().split("$$$$\n")
approved = []
for block in blocks:
    parts = block.split("> <DRUGBANK_ID>\n")
    tag = parts[1].split("\n")[0] if len(parts) > 1 else None
    approved.append(tag in db_approved_ids)
print(sum(approved), len(approved))
```
