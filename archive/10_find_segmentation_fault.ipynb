```python
import pandas as pd
```

```python
with open("log.txt") as file:
    lines = sorted(file.read().strip().split("\n"))
df = pd.DataFrame(lines, columns=["name"])
```

```python
df["file"] = df["name"].str.split("/").str[0].str.split("_").str[-1].apply(int)
df["mol"] = df["name"].str.split("/").str[1].str.split(".").str[0].str.split("_").str[-1].apply(int)
```

```python
df
```

```python
tuples = set(zip(df["file"], df["mol"]))
tuples
```

```python
missing = set()
for f in [0, 1, 2]:
    for i in range(10000):
        if (f, i) not in tuples:
            missing.add((f, i))
```
