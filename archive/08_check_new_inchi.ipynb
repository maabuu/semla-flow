```python
import rdkit
from rdkit.Chem.inchi import MolFromInchi, MolToInchi
from rdkit.Chem.rdmolfiles import MolToSmiles, SDMolSupplier
from tqdm import tqdm
```

```python
"RDKit: " + rdkit.__version__
```

```python
data = "/homes/buttensc/Projects/semla-flow/data/unconditional/geom-drugs/test.sdf"
mols = list(tqdm(SDMolSupplier(data)))
```

```python
inchis = [MolToInchi(mol) for mol in mols]
```

```python
smiles = [MolToSmiles(mol) for mol in mols]
```

```python
len(set(smiles))
```

```python
inchis_short = [inchi.split("=1S/")[1].split("/h")[0] for inchi in inchis]
```
