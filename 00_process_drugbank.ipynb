```python
import os
import sys
from pathlib import Path

import pandas as pd
```

```python
dir = Path("/vols/opig/users/buttensc/Storage/Projects/3d_mol_gen_benchmark")
os.chdir(dir)
```

```python
# meta data
df = pd.read_csv("data/unconditional/drugbank/all_structure_links.csv")
df["approved"] = df["Drug Groups"].str.replace("vet_approved", "").str.lower().str.contains("approve")
all_approved_ids = set(df[df["approved"]]["DrugBank ID"])
print(f"Number of approved drugs: {sum(df.approved)}, {len(all_approved_ids)}")
```

```python
file = Path("data/unconditional/drugbank/all_structures_3d.sdf")
blocks = open(file).read().strip("$$$$").strip().split("$$$$\n")
df = pd.DataFrame(blocks, columns=["block"])
df["id"] = df["block"].str.split("> <DRUGBANK_ID>\n").str[1].str.split("\n").str[0].str.strip()
df = df.sort_values("id")
df["approved"] = df["id"].isin(all_approved_ids)
print(f"Number of approved drugs: {sum(df.approved)} out of {len(df)}")
```

```python
file = "/vols/opig/users/buttensc/Storage/Projects/3d_mol_gen_benchmark/data/unconditional/drugbank/approved_structures_3d.sdf"
with open(file, "w") as f:
    for block in df[df["approved"]]["block"]:
        f.write(block + "$$$$\n")
```
