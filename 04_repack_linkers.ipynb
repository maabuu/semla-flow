```python
from pathlib import Path

from rdkit.Chem.rdmolfiles import SDMolSupplier, SDWriter
from rdkit.Chem.rdmolops import AddHs, SanitizeMol
```

```python
input_file = (
    "/homes/buttensc/Projects/semla-flow/data/conditional_fragments/reference_mols.sdf"
)
output_file = "/homes/buttensc/Projects/semla-flow/data/conditional_fragments/reference_mols_named.sdf"
```

```python
supplier = SDMolSupplier(input_file, sanitize=False, removeHs=False)
writer = SDWriter(output_file)

for i, mol in enumerate(supplier):
    if mol is None:
        continue
    mol.SetProp("_Name", f"mol_{i}")
    writer.write(mol)
```
