```python
from pathlib import Path

from rdkit.Chem.AllChem import ReplaceSubstructs
from rdkit.Chem.rdmolfiles import MolFromSmarts, SDMolSupplier
from rdkit.Chem.rdmolops import SanitizeMol
```

```python
file = Path("data/conditional_fragments/reference_mols.sdf")
supplier = SDMolSupplier(str(file), sanitize=False, strictParsing=True)
```

```python
for i, mol in enumerate(supplier):
    mol.RemoveAllConformers()
    # if SanitizeMol(mol, catchErrors=True) != 0:
    if i == 216:
        break
```

```python
mol.GetSubstructMatch(MolFromSmarts("C1~N~C~N~C~N~1"))
mol
```

```python
ReplaceSubstructs(
    mol,
    MolFromSmarts("C1~N~C~N~C~N~1"),
    MolFromSmarts("C1=N-C=N-C=N-1"),
    replaceAll=True,
)[0]
```

```python
SanitizeMol(mol, catchErrors=True)
```
